# Flight animation: the reachable tubes and the GPs

A real-time video of one flight (MP4, plus a lighter GIF preview), one frame every 50 ms, so every plan is on screen
while it is in use:

* **whole flight** (left): the path, the current plan's tube, a faint trail of every earlier certified tube, the goal
  and the ground floor; the box shows the follow-cam window.
* **follow-cam** (right): the current plan's mixed-monotone tube drawn as what it is, one interval box per rollout
  step, colored by look-ahead time (rows the vehicle has already passed are faded). The certificate is spatial: it ends
  at the first box whose y or altitude bound is more than the threshold (0.5 m) from the reference position, or that
  reaches the floor. That box is outlined in red, the dashed square is the reference +- threshold at that step, the
  hatched boxes are the uncertified margin kept after it, and the text gives the reason.
* **wind GPs** (bottom): the y-wind (a function of altitude) and z-wind (a function of y) *exactly as that plan used
  them*, mean +- 3 sigma, observations sized by recency; the shaded span is the range the certified tube covers.
