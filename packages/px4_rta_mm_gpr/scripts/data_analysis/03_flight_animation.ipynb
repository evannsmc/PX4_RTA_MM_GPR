# Flight animation with the GPs

Animated GIF of one flight: the vehicle, its flown path, the current plan's certified tube (lower/upper bounds) and
reference, and the two wind GPs *exactly as that plan used them* (reconstructed from the GP data stored with the plan):
the y-wind as a function of altitude and the z-wind as a function of lateral position, mean +- 3 sigma, with the
observations (older ones smaller).
